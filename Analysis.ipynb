## Data Analysis of Titanic Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tested.csv')
```

```python
df.head()
```

```python
df.Age.mean()
```

# Observation :
The average age of travellers on titanic was 30.27

```python
traveller = df['Sex'].value_counts()
```

```python
plt.pie(traveller,labels=['Males','Females'],autopct='%1.2f%%')
plt.title('Traveller based on gender')
```

# Observation: 
Almost 64% of Passenger were Males

```python
total = df.groupby('Sex')['Survived'].sum()
```

```python
plt.bar(['Females','Males'],total)
plt.title('Survived')
plt.ylabel('Survival Count')
plt.xlabel('Gender')
```

# Observation :
No men Survived during the accident of Titanic

```python
class_wise = df.groupby('Pclass')['Survived'].mean()
```

```python
plt.pie(class_wise,labels=['First class','Second Class','Third Class'],autopct='%1.2F%%')
plt.title('Survival Based on Class')
```

# Observation :
Almost half of the survivers are from First class

```python

age_wise = pd.cut(df['Age'], bins=[0, 12, 18, 60, 100],labels=['Child', 'Teen', 'Adult', 'Senior'])
```

```python
plt.bar(['Adult','Teen','Child','Senior'],age_wise.value_counts())
```

# Observation:
Most passenger in Titanic where Adults that is age group of 19-60
